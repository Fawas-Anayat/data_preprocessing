## scaling

### standard scaler
sets the values such that their mean is at zero and the standard deviation is at one

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import RobustScaler
```

```python
df=pd.read_csv('data\Fortune 500 Companies.csv')
```

```python
df.head(2)
```

```python
df.info()
```

```python
df.shape
```

```python
a=df['revenue_mil']
a=pd.DataFrame(a)
```

```python
a.head()
type(a)
```

```python
a.describe()
```

```python
scaler=StandardScaler()
scaler.fit(a)
a_new=scaler.transform(a)
```

```python
a_new_df=pd.DataFrame(a_new)
```

```python
a_new_df.head()
```

```python
a_new_df.describe()
```

```python
a.plot(kind='kde')
```

```python
a_new_df.plot(kind='kde')
```

## normalisation
### minmax scaler

```python
a_minmax=df['revenue_mil']
a_minmax_df=pd.DataFrame(a_minmax)
scaler=MinMaxScaler()
a_minmax_df_scaled=pd.DataFrame(scaler.fit_transform(a_minmax_df))
```

```python
a_minmax_df_scaled.head()
```

```python
a_minmax_df_scaled.describe()
```

```python
a_minmax_df_scaled.plot(kind='density')
```
